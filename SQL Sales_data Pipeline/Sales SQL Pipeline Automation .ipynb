```python
import pandas as pd
import psycopg2
import time
import os
```

```python
# Configurations
csv_file = r"E-Commerce Data.csv"
db_params = {
    "dbname": "Sales Pipeline",
    "user": "postgres",
    "password": "your Password",
    "host": "localhost",
    "port": "5432",
}

# Get initial file modification time
last_modified = os.path.getmtime(csv_file)

while True:
    time.sleep(20)  # Check every 20 seconds
    new_modified = os.path.getmtime(csv_file)

    if new_modified != last_modified:
        print("CSV updated! Inserting new data into PostgreSQL...")

        # Load CSV
        df = pd.read_csv(csv_file)

        # convert CustomerID, since postgre expects customerId as int value but pandas considers this as float (default)
        df["CustomerID"] = df["CustomerID"].fillna(0).astype("Int64")

        # Convert DataFrame to list of tuples for bulk insertion
        data = df.to_records(index=False).tolist()

        # Connect to PostgreSQL
        conn = psycopg2.connect(**db_params)
        cur = conn.cursor()

        # Bulk insertion query
        query = """INSERT INTO staging_sales (invoiceno, stockcode, description, quantity, invoicedate, unitprice, customerid, country) 
                   VALUES (%s, %s, %s, %s, %s, %s, %s, %s)"""

        # Use executemany for efficient bulk insert
        cur.executemany(query, data)
        conn.commit()
        cur.close()
        conn.close()
        
        last_modified = new_modified
        print("Data successfully inserted!")
    else:
        print("No New Update in CSV")
```
